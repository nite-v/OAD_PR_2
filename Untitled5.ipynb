1. Загрузка данных

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("train.csv")
```

2. Знакомство с датасетом

```python
print("Количество строк и столбцов:", df.shape)
print("\nНазвания столбцов:")
print(df.columns.tolist())
print("\nТипы данных:")
print(df.dtypes)
print("\nИнформация о датасете:")
print(df.info())
```

3. Анализ пассажиров

```python
# Количество пассажиров каждого класса
print("\nКоличество пассажиров по классам:")
print(df["Pclass"].value_counts().sort_index())

# Количество мужчин и женщин
print("\nКоличество мужчин и женщин:")
print(df["Sex"].value_counts())

# Разделение на возрастные группы
bins = [0, 12, 18, 30, 45, 60, 100]
labels = ["0-12", "13-18", "19-30", "31-45", "46-60", "60+"]
df["AgeGroup"] = pd.cut(df["Age"], bins=bins, labels=labels, right=True)

print("\nКоличество пассажиров по возрастным группам:")
print(df["AgeGroup"].value_counts().sort_index())

# Средний возраст
print("\nСредний возраст пассажиров:", df["Age"].mean())

# Медианный возраст
print("Медианный возраст пассажиров:", df["Age"].median())

# Средняя стоимость билета
print("Средняя стоимость билета:", df["Fare"].mean())
```

4. Сравнение пассажиров разных классов

```python
# Средняя стоимость билета по классам
print("\nСредняя стоимость билета по классам:")
print(df.groupby("Pclass")["Fare"].mean())

# Средний возраст по классам
print("\nСредний возраст по классам:")
print(df.groupby("Pclass")["Age"].mean())

# Количество мужчин и женщин в каждом классе
print("\nКоличество мужчин и женщин по классам:")
print(pd.crosstab(df["Pclass"], df["Sex"]))

# В каком классе было больше всего пассажиров
max_class = df["Pclass"].value_counts().idxmax()
print("\nБольше всего пассажиров было в классе:", max_class)
```

Вывод по классам: В 1 классе средняя стоимость билета и средний возраст выше, чем в 3 классе. Доля женщин в 1 классе выше, чем в 3 классе, хотя абсолютное число женщин больше именно в 3 классе.

```python
# Самый дорогой билет
max_fare_row = df.loc[df["Fare"].idxmax()]
print("\nСамый дорогой билет:")
print(max_fare_row[["Name", "Pclass", "Age", "Sex", "Fare", "Survived"]])
```

5. Анализ выживаемости

```python
# Общее количество выживших и погибших
surv_counts = df["Survived"].value_counts().rename({0: "Погибшие", 1: "Выжившие"})
print("\nОбщее количество выживших и погибших:")
print(surv_counts)

# Выживаемость мужчин и женщин
print("\nДоля выживших по полу:")
print(df.groupby("Sex")["Survived"].mean())

# Выживаемость мужчин и женщин для каждого класса
print("\nДоля выживших по классу и полу:")
surv_sex_class = df.groupby(["Pclass", "Sex"])["Survived"].mean().unstack()
print(surv_sex_class)

# Группа с наибольшей долей выживших
max_group = df.groupby(["Pclass", "Sex"])["Survived"].mean().idxmax()
max_value = df.groupby(["Pclass", "Sex"])["Survived"].mean().max()

print(f"\nНаибольшая доля выживших: класс {max_group[0]}, пол {max_group[1]}, доля = {max_value:.2%}")
```

6. Визуализация

```python
# График количества пассажиров по классам
plt.figure(figsize=(6, 4))
df["Pclass"].value_counts().sort_index().plot(kind="bar")
plt.title("Количество пассажиров по классам")
plt.xlabel("Класс")
plt.ylabel("Количество пассажиров")
plt.xticks(rotation=0)
plt.show()

# График выживаемости по полу
plt.figure(figsize=(6, 4))
df.groupby("Sex")["Survived"].mean().plot(kind="bar")
plt.title("Доля выживших по полу")
plt.xlabel("Пол")
plt.ylabel("Доля выживших")
plt.xticks(rotation=0)
plt.show()
```

Собственный вопрос к данным: зависит ли выживаемость от размера семьи пассажира?

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

family_surv = df.groupby("FamilySize")["Survived"].mean()
print("\nДоля выживших по размеру семьи:")
print(family_surv)

plt.figure(figsize=(8, 4))
family_surv.plot(kind="bar")
plt.title("Доля выживших в зависимости от размера семьи")
plt.xlabel("Размер семьи")
plt.ylabel("Доля выживших")
plt.xticks(rotation=0)
plt.show()
```

Пассажиры, путешествовавшие в одиночку, выживали реже, чем семьи из 2–4 человек. Семьи из 5 и более человек также имели низкую долю выживших, за исключением отдельных случаев.
